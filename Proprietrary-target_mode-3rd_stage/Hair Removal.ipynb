## Hair Preprocessing - Create `skin_clean_R_G_IR_aligned_1_hairless`

```python
import sys
sys.path.insert(0, "/home/i_am_helium/my_libs")
import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Per-channel parameters
CHANNEL_PARAMS = {
    'R': {
        'kernel_size': (9, 9),
        'threshold': 6,
        'dilate_iterations': 2,
        'min_hair_pixels': 2500,
        'inpaint_radius': 5
    },
    'G': {
        'kernel_size': (9, 9),
        'threshold': 16,
        'dilate_iterations': 2,
        'min_hair_pixels': 3000,
        'inpaint_radius': 5,
        'remove_small_objects': 150
    },
    'IR': {
        'kernel_size': (11, 11),
        'threshold': 25,
        'dilate_iterations': 2,
        'min_hair_pixels': 7000,
        'inpaint_radius': 6,
        'remove_small_objects': 200
    }
}

def get_hair_mask_per_channel(img_path, channel_name):
    params = CHANNEL_PARAMS[channel_name]
    
    img = cv2.imread(img_path)
    if img is None:
        return None, 0, False
    
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    # Blackhat
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, params['kernel_size'])
    blackhat = cv2.morphologyEx(gray, cv2.MORPH_BLACKHAT, kernel)
    
    # Threshold
    _, mask = cv2.threshold(blackhat, params['threshold'], 255, cv2.THRESH_BINARY)
    
    # NEW: Morphological opening to remove small noise
    if params.get('use_morphological_opening', False):
        opening_kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
        mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, opening_kernel)
    
    # NEW: Remove small disconnected components
    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(mask, connectivity=8)
    
    cleaned_mask = np.zeros_like(mask)
    min_size = params.get('remove_small_objects', 100)
    
    for i in range(1, num_labels):
        area = stats[i, cv2.CC_STAT_AREA]
        if area >= min_size:
            cleaned_mask[labels == i] = 255
    
    # Dilate after cleaning
    cleaned_mask = cv2.dilate(cleaned_mask, None, iterations=params['dilate_iterations'])
    
    hair_pixels = np.sum(cleaned_mask > 0)
    should_inpaint = hair_pixels >= params['min_hair_pixels']
    
    return cleaned_mask, hair_pixels, should_inpaint

def process_hair_removal_per_channel(row, output_base_path, visualize=False):
    """
    Process hair removal for one lesion, each channel independently.
    """
    lesion_name = row['lesion_name']
    class_name = row['class']
    
    results = {}
    
    for channel in ['R', 'G', 'IR']:
        img_path = row[f'{channel}_path']
        
        # Read original image
        img = cv2.imread(img_path)
        if img is None:
            print(f"Cannot read {channel}: {img_path}")
            continue
        
        # Detect hair mask for this channel
        mask, hair_pixels, should_inpaint = get_hair_mask_per_channel(img_path, channel)
        
        # Apply inpainting if needed
        if should_inpaint:
            params = CHANNEL_PARAMS[channel]
            cleaned = cv2.inpaint(img, mask, params['inpaint_radius'], cv2.INPAINT_TELEA)
            decision = f"HAIR REMOVED ({hair_pixels} pixels)"
        else:
            cleaned = img.copy()
            decision = f"NO SIGNIFICANT HAIR ({hair_pixels} pixels)"
        
        results[channel] = {
            'original': img,
            'mask': mask,
            'cleaned': cleaned,
            'hair_pixels': hair_pixels,
            'processed': should_inpaint,
            'decision': decision
        }
        
        # Save cleaned image
        output_dir = Path(output_base_path) / class_name / lesion_name
        output_dir.mkdir(parents=True, exist_ok=True)
        output_path = output_dir / f'{channel}.png'
        cv2.imwrite(str(output_path), cleaned)
    
    # Visualization
    if visualize:
        fig, axes = plt.subplots(3, 3, figsize=(15, 15))
        fig.suptitle(f'{class_name} | {lesion_name}', fontsize=16)
        
        for row_idx, channel in enumerate(['R', 'G', 'IR']):
            if channel not in results:
                continue
            
            r = results[channel]
            
            # Original
            img_rgb = cv2.cvtColor(r['original'], cv2.COLOR_BGR2RGB)
            axes[row_idx, 0].imshow(img_rgb)
            axes[row_idx, 0].set_title(f"{channel} original")
            axes[row_idx, 0].axis("off")
            
            # Mask
            axes[row_idx, 1].imshow(r['mask'], cmap="gray")
            axes[row_idx, 1].set_title(f"{channel} mask\n{r['decision']}")
            axes[row_idx, 1].axis("off")
            
            # Cleaned
            cleaned_rgb = cv2.cvtColor(r['cleaned'], cv2.COLOR_BGR2RGB)
            axes[row_idx, 2].imshow(cleaned_rgb)
            axes[row_idx, 2].set_title(f"{channel} cleaned")
            axes[row_idx, 2].axis("off")
        
        plt.tight_layout()
        plt.show()
    
    return results

# ===== USAGE =====

# Load dataset structure
csv_path = "/home/i_am_helium/skin_clean_R_G_IR_aligned_1/dataset_structure.csv"
df = pd.read_csv(csv_path)

# Add full paths
base_path = "/home/i_am_helium/skin_clean_R_G_IR_aligned_1"
for channel in ['R', 'G', 'IR']:
    df[f'{channel}_path'] = df[f'{channel}_path'].apply(lambda x: os.path.join(base_path, x))

# ===== PROCESS ALL LESIONS =====
print(f"Processing ALL {len(df)} lesions...")
print("=" * 100)

for idx, row in df.iterrows():
    if (idx + 1) % 50 == 0:
        print(f"Processed {idx + 1}/{len(df)} lesions...")
    
    process_hair_removal_per_channel(
        row, 
        output_base_path="/home/i_am_helium/skin_clean_R_G_IR_aligned_1_hairless",
        visualize=False
    )

print(f"\nDone! Processed all {len(df)} lesions.")
print(f"Output saved to: /home/i_am_helium/skin_clean_R_G_IR_aligned_1_hairless")

# ===== VISUALIZE 10 RANDOM SAMPLES =====
print("\n" + "=" * 100)
print("Now showing 10 random samples with visualization...")
print("=" * 100)

SEED = 52
np.random.seed(SEED)
sample_df = df.sample(n=10, random_state=SEED).reset_index(drop=True)

for idx, row in sample_df.iterrows():
    print(f"\n{idx+1}/10 | {row['class']} | {row['lesion_name']}")
    results = process_hair_removal_per_channel(
        row, 
        output_base_path="/home/i_am_helium/skin_clean_R_G_IR_aligned_1_hairless",
        visualize=True
    )
    
    # Print summary
    for channel in ['R', 'G', 'IR']:
        if channel in results:
            r = results[channel]
            print(f"  {channel}: {r['decision']}")
```
